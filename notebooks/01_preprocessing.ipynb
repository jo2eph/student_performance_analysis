## Data Cleaning

In this section, we are going to clean the dataset before we start performing any data analysis.

We will examine the dataset for any **missing values** or **duplicates**.
For this process, we will leave out outliers is there is any.

### Missingness

Let's start with missing values. First, let's see how many missing values this dataset contains.

```python
n_missing_values = students.isna().sum().sum()

print(f"This dataset has {n_missing_values} missing value(s).")

del n_missing_values
```

As we can see from the above output, there are missing values present in this dataset.

Next, let's examine what column(s) contain missing values.

```python
students.isna().sum()
```

From the above, we can see that there appears to be only one column that has missing values: `favorite_AI_tool`.

```python
# Let's take a look at the unique values of `favorite_AI_tool`
students['favorite_AI_tool'].unique()
```

Notice that this column is **missing by design**, meaning that the missing values are intentional, since not every student have a favorite AI tool that they prefer using.

Now, there are a few approaches we can take to handle missingness here.

One way we can handle it is to drop the column `favorite_AI_tool`, or drop the rows that contain a missing value.

However, the approach we are going to use is to actually keep the column, and replace the null entries with a string value like 'None'.
To do that, we are going to use the `.fillna()` method to fill the missing values for the column.

```python
# Fill missing values for column `favorite_AI_tool`
students['favorite_AI_tool'] = students['favorite_AI_tool'].fillna("None")
students
```

```python
# Checking to make sure missing values are filled
print(f"There are {students.isna().sum().sum()} missing values.")
```

### Duplicates

The next step is to find out whether there are duplicates present, and handle duplicated data.

The simplest method to do that is to use the `df.duplicated()` method.

```python
print(f"There are {students.duplicated().sum()} duplicated rows.")
```

Since there are no duplicated rows, we are done.

Note that if there were duplicate rows, we can remove them via the `df.drop_duplicates()` method.

### Save Cleaned Data

Now that we are finished with cleaning the data, let's save the dataset to a CSV file using `.to_csv()`.

We are going to save it into the `../data/processed/` directory, save the data as `students_cleaned.csv`.

```python
students.to_csv("../data/processed/students_cleaned.csv", index=False)
```
