## EDA - Univariate Analysis

Next, we are going to perform exploratory analysis to explore our dataset, uncover hidden patterns, and get to know our dataset better.

```python
def plot_hist(feature):
    plt.figure();
    plt.hist(students[feature]);
    plt.xlabel(feature);
    plt.ylabel('Count');
    plt.title(f'Distribution of {feature}');
```

```python
plot_hist('GPA')
```

From the figure above, `GPA` is skewed to the left. The mean of `GPA` is 3.618.

Let's see the distribution of hours students study per day, `study_hours_per_day`.

```python
plot_hist('study_hours_per_day')
```

```python
plot_hist('class_attendance_percent')
```

```python
plot_hist('sleep_hours')
```

```python
plot_hist('screen_time_hours')
```
